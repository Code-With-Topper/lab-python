# Practical 9 — Sentiment Analysis Using NLP

## Objective
Perform a simple sentiment-analysis workflow using NLTK's VADER lexicon on example sentences or user-provided text.

**How to use:** Put the required CSV file(s) in the `data/` folder, update the filename/path if needed, and run cells from top to bottom. Where a dataset is unavailable, the notebook reports that clearly rather than inventing results.

## Imports

```python
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

DATA_DIR = Path("../data")
DATA_DIR.mkdir(exist_ok=True)
```

## Setup VADER

```python
import nltk
try:
    from nltk.sentiment import SentimentIntensityAnalyzer
    nltk.data.find("sentiment/vader_lexicon.zip")
except LookupError:
    nltk.download("vader_lexicon")
    from nltk.sentiment import SentimentIntensityAnalyzer
```

## Analyze sample text

```python
sia = SentimentIntensityAnalyzer()
samples = [
    "The product is excellent and I am very happy with the service.",
    "The delivery was slow and the experience was disappointing.",
    "The item arrived yesterday.",
    "I absolutely love this! It works wonderfully.",
    "This is the worst experience I have had."
]
def label_sentiment(compound):
    if compound >= 0.05:
        return "Positive"
    if compound <= -0.05:
        return "Negative"
    return "Neutral"

results = []
for sentence in samples:
    scores = sia.polarity_scores(sentence)
    results.append({"text": sentence, **scores, "sentiment": label_sentiment(scores["compound"])})
sentiment_df = pd.DataFrame(results)
display(sentiment_df)
```

## Analyze your own text

```python
user_text = input("Enter a sentence or review: ").strip()
if user_text:
    scores = sia.polarity_scores(user_text)
    print("Scores:", scores)
    print("Sentiment:", label_sentiment(scores["compound"]))
else:
    print("No text entered.")
```

## Optional: visualize example sentiment

```python
sentiment_counts = sentiment_df["sentiment"].value_counts()
sentiment_counts.plot(kind="bar", edgecolor="black", figsize=(6, 4))
plt.xlabel("Sentiment")
plt.ylabel("Number of example sentences")
plt.title("Sentiment labels for sample text")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

## Conclusion

VADER is a lightweight lexicon-based baseline. It can misread sarcasm, context, domain-specific language, and mixed sentiment; treat the labels as estimates, not ground truth.
