# Practical 2 — Customer Churn Data Analysis

## Objective
Count duplicate CustomerID records, missing TotalCharges values, average MonthlyCharges, malformed Dependents values `1@#`, and the tenure data type.

**How to use:** Put the required CSV file(s) in the `data/` folder, update the filename/path if needed, and run cells from top to bottom. Where a dataset is unavailable, the notebook reports that clearly rather than inventing results.

## Imports

```python
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

DATA_DIR = Path("../data")
DATA_DIR.mkdir(exist_ok=True)
```

```python
csv_path = DATA_DIR / "Churn.csv"  # Change to the actual supplied filename if different.
if not csv_path.exists():
    raise FileNotFoundError("Place the churn dataset in data/Churn.csv (or update csv_path).")
churn = pd.read_csv(csv_path)
display(churn.head())
print("Columns:", churn.columns.tolist())
```

```python
# Normalize column-name matching without changing the original dataframe.
def find_col(df, candidates):
    lookup = {str(c).strip().lower(): c for c in df.columns}
    for name in candidates:
        if name.lower() in lookup:
            return lookup[name.lower()]
    raise KeyError(f"Could not find any of these columns: {candidates}")

customer_col = find_col(churn, ["customerID", "customer_id"])
total_col = find_col(churn, ["TotalCharges", "total_charges"])
monthly_col = find_col(churn, ["MonthlyCharges", "monthly_charges"])
dependents_col = find_col(churn, ["Dependents"])
tenure_col = find_col(churn, ["tenure"])
```

```python
print("Duplicate records based on CustomerID:", int(churn[customer_col].duplicated().sum()))
print("Missing values in TotalCharges:", int(churn[total_col].isna().sum()))
total_numeric = pd.to_numeric(churn[total_col].astype(str).str.strip().replace("", np.nan), errors="coerce")
print("TotalCharges blank/non-numeric values after parsing:", int(total_numeric.isna().sum()))
monthly_numeric = pd.to_numeric(churn[monthly_col], errors="coerce")
print("Average MonthlyCharges:", monthly_numeric.mean())
print("Dependents values equal to '1@#':", int(churn[dependents_col].astype(str).eq("1@#").sum()))
print("tenure data type:", churn[tenure_col].dtype)
```

Record the output values above. Missing values detected after numeric conversion include blanks or non-numeric strings, which may differ from `isna()` on the raw CSV.
