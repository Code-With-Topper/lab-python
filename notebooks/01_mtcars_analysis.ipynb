# Practical 1 — mtcars Dataset Analysis

## Objective
Read the dataset, inspect its first rows and column types, plot the `mpg` histogram, identify the most frequent histogram interval, and interpret `mpg` versus `wt`.

**How to use:** Put the required CSV file(s) in the `data/` folder, update the filename/path if needed, and run cells from top to bottom. Where a dataset is unavailable, the notebook reports that clearly rather than inventing results.

## Imports

```python
from pathlib import Path
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

DATA_DIR = Path("../data")
DATA_DIR.mkdir(exist_ok=True)
```

## 1. Read the dataset

```python
csv_path = DATA_DIR / "mtcars.csv"
if csv_path.exists():
    mtcars = pd.read_csv(csv_path)
else:
    # Optional fallback: seaborn's bundled dataset (if seaborn/data access is available).
    try:
        import seaborn as sns
        mtcars = sns.load_dataset("mpg").rename(columns={"weight": "wt"})
        if "mpg" not in mtcars.columns:
            raise ValueError("Fallback dataset does not contain mpg.")
        print("Loaded seaborn mpg dataset as a fallback. For exact lab results, use the supplied mtcars.csv.")
    except Exception as exc:
        raise FileNotFoundError("Place mtcars.csv in the data/ folder. The PDF does not include the dataset.") from exc
mtcars.head()
```

## 2. Head and data types

```python
display(mtcars.head())
print("\nColumn data types:")
display(mtcars.dtypes.to_frame("dtype"))
```

## 3. Histogram of mpg and most frequent interval

```python
if "mpg" not in mtcars.columns:
    raise KeyError("Expected a column named 'mpg'.")
mpg = pd.to_numeric(mtcars["mpg"], errors="coerce").dropna()
counts, edges = np.histogram(mpg, bins="auto")
idx = int(np.argmax(counts))
plt.figure(figsize=(8, 5))
plt.hist(mpg, bins="auto", edgecolor="black")
plt.xlabel("Miles per gallon (mpg)")
plt.ylabel("Frequency")
plt.title("Distribution of mpg")
plt.tight_layout()
plt.show()
print(f"Highest frequency: {counts[idx]} observations")
print(f"Corresponding interval: [{edges[idx]:.3g}, {edges[idx+1]:.3g}{']' if idx == len(counts)-1 else ')'}")
```

## 4. Scatter plot: mpg versus wt

```python
if not {"mpg", "wt"}.issubset(mtcars.columns):
    raise KeyError("Expected columns 'mpg' and 'wt'. If your file uses 'weight', rename it to 'wt'.")
plt.figure(figsize=(8, 5))
plt.scatter(mtcars["wt"], mtcars["mpg"], alpha=0.8)
plt.xlabel("Weight (wt)")
plt.ylabel("Miles per gallon (mpg)")
plt.title("mpg versus wt")
plt.tight_layout()
plt.show()
print("Pearson correlation:", mtcars[["mpg", "wt"]].corr().iloc[0, 1])
print("Interpretation: a negative correlation indicates that heavier cars tend to have lower mpg; inspect the plot and correlation for this dataset.")
```
